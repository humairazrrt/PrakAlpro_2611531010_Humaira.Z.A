## 2.1 Dasar Teori

### a. Python

Python merupakan bahasa pemrograman tingkat tinggi yang memiliki sintaks sederhana dan mudah dipahami. Python mendukung berbagai konsep pemrograman, salah satunya adalah perulangan.

### b. Perulangan `for`

Perulangan `for` digunakan untuk menjalankan suatu perintah secara berulang berdasarkan urutan atau rentang nilai tertentu. Perulangan ini banyak digunakan untuk mengolah data dan membuat pola.

### c. Fungsi `range()`

Fungsi `range()` digunakan untuk menentukan rentang nilai dalam perulangan `for`. Contohnya `range(1, 6)` menghasilkan angka 1 sampai 5.

### d. Nested Loop

*Nested loop* adalah perulangan yang berada di dalam perulangan lainnya. Konsep ini dapat digunakan untuk membuat pola angka, karakter, simbol, maupun bentuk yang terdiri dari baris dan kolom.

### e. Fungsi `print()`

Fungsi `print()` digunakan untuk menampilkan hasil program. Dalam perulangan, `end=""` dapat digunakan agar hasil tetap dicetak pada baris yang sama, sedangkan `print()` digunakan untuk berpindah ke baris berikutnya.

### 2.2.7 Program "nested_for4_2611531010.py"

````python
# Buat file dengan nama nested_for4_NIM.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

tinggi_1010 = int(input("Masukkan tinggi pola (bilangan genap, misal 10): "))

if tinggi_1010 % 2 != 0:
    print("Tinggi harus bilangan genap!")
else:
    a_1010 = tinggi_1010
    c_1010 = a_1010
    lebar_1010 = (2 * tinggi_1010) - 2

    for i_1010 in range(1, tinggi_1010 + 1):
        b_1010 = c_1010 + 1

        for j_1010 in range(1, lebar_1010 + 1):

            # Baris atas dan bawah
            if i_1010 == 1 or i_1010 == tinggi_1010:
                if j_1010 == 1 or j_1010 == lebar_1010:
                    print("#", end="")
                else:
                    print("=", end="")

            # Baris isi
            else:
                if j_1010 == 1 or j_1010 == lebar_1010:
                    print("|", end="")
                elif j_1010 == c_1010:
                    print("<", end="")
                elif j_1010 == b_1010:
                    print(">", end="")
                elif j_1010 == (lebar_1010 - c_1010):
                    print("<", end="")
                elif j_1010 == (lebar_1010 - c_1010 + 1):
                    print(">", end="")
                elif b_1010 < j_1010 < (lebar_1010 - c_1010):
                    print(".", end="")
                else:
                    print(" ", end="")

        print()  # pindah baris setelah satu baris selesai

        # Logika asli Java (update nilai untuk baris berikutnya)
        a_1010 -= 2
        if a_1010 <= 0:
            c_1010 = (-a_1010) + 2
        else:
            c_1010 = a_1010                
````

````python
# Buat file dengan nama jumlah_genap_NIM.py
# Buat program untuk perulangan for dalam Python
# Nama variabel ditambah 4 digit nim terakhir contoh: ulang_1234
# Program ini menggunakan fungsi input()

ulang_1010 = int(input("Masukkan nilai batas: "))

jumlah_1010 = 0
for i_1010 in range(1, ulang_1010 + 1):
    if i_1010 % 2 == 0:
        print(i_1010, end="")
        jumlah_1010 = jumlah_1010 + i_1010
        
        if i_1010 < ulang_1010:
            print("+", end="")
        else:
            print("=", jumlah_1010, end="")
print()
print("Jumlah =", jumlah_1010)
````
